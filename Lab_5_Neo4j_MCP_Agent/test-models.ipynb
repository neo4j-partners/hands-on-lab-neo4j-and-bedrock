# Test Bedrock Model Setup

This notebook tests the Bedrock model configuration to verify the model ARN and API setup is working correctly.

```python
SERVICE_NAME = 'bedrock-runtime'
REGION_NAME = 'us-west-2'

# Application Inference Profile ARN (required for SageMaker Unified Studio)
INFERENCE_PROFILE_ARN = 'arn:aws:bedrock:us-west-2:ACCOUNT_ID:application-inference-profile/PROFILE_ID'
```

```python
import boto3
import json

bedrock = boto3.client(
    service_name=SERVICE_NAME,
    region_name=REGION_NAME
)
```

```python
import traceback
def call_language_model(prompt_data):
    try:
        body = json.dumps({
            "anthropic_version": "bedrock-2023-05-31",
            "max_tokens": 2048,
            "temperature": 0,
            "top_k": 1,
            "top_p": 0.88,
            "messages": [
                {
                    "role": "user",
                    "content": prompt_data
                }
            ]
        })

        response = bedrock.invoke_model(
            body=body, 
            modelId=INFERENCE_PROFILE_ARN,
            accept='application/json',    
            contentType='application/json'
        )

        response_body = json.loads(response.get('body').read())
        return response_body['content'][0]['text']

    except Exception as e:
        print(f"Error: {e}") 
        print(traceback.format_exc())
        return None
```

## Test the Model

Send a simple test prompt to verify the model is working.

```python
response = call_language_model("Say hello in one sentence.")
print(response)
```
