# Test Bedrock Model Setup

This notebook tests the Bedrock model configuration to verify the model ARN and API setup is working correctly.

```python
SERVICE_NAME = 'bedrock-runtime'
REGION_NAME = 'us-east-1'
```

```python
import boto3
import json
bedrock = boto3.client(
 service_name=SERVICE_NAME,
 region_name=REGION_NAME,
 endpoint_url=f'https://{SERVICE_NAME}.{REGION_NAME}.amazonaws.com'
)
```

```python
import traceback
def call_language_model(prompt_data):
    try:
        # Use Messages API format (required for Claude Sonnet 4)
        body = json.dumps({
            "anthropic_version": "bedrock-2023-05-31",
            "max_tokens": 2048,
            "temperature": 0,
            "top_k": 1,
            "top_p": 0.88,
            "messages": [
            {
                "role": "user",
                "content": prompt_data
            }
            ]
        })
        # Correct inference profile ID
        modelId = 'us.anthropic.claude-sonnet-4-20250514-v1:0'
        accept = 'application/json'    
        contentType = 'application/json'

        response = bedrock.invoke_model(
            body=body, 
            modelId=modelId, 
            accept=accept, 
            contentType=contentType
        )

        response_body = json.loads(response.get('body').read())

        # Messages API returns content in a different structure
        return response_body['content'][0]['text']

    except Exception as e:
        print(f"Error: {e}") 
        print(traceback.format_exc())
        return None
```

## Test the Model

Send a simple test prompt to verify the model is working.

```python
response = call_language_model("Say hello in one sentence.")
print(response)
```
